### Note about phone numbers

During the initial data check, I noticed that the `PHONE` column contains values in different formats. Some numbers are stored in an international format, some in a local format, and some contain symbols such as spaces, dots, brackets, or country codes.

Because of this, the `PHONE` column is not fully standardized and should not be used as a reliable customer identifier without additional cleaning and validation.

However, this issue does not affect the RFM analysis, because RFM segmentation is based on customer purchase behavior, not phone numbers. For this analysis, customers are identified by `CUSTOMERNAME`, and the key fields are:

- `CUSTOMERNAME`
- `ORDERNUMBER`
- `ORDERDATE`
- `SALES`

Therefore, the inconsistent phone number formatting is documented as a data quality issue, but it is not critical for the RFM analysis.

```python
# Берем только нужные колонки для RFM-анализа
rfm_df = df[["CUSTOMERNAME", "ORDERNUMBER", "ORDERDATE", "SALES"]].copy()

# На всякий случай проверяем, что дата в формате datetime
rfm_df["ORDERDATE"] = pd.to_datetime(rfm_df["ORDERDATE"], errors="coerce")

# Дата анализа — один день после последнего заказа
snapshot_date = rfm_df["ORDERDATE"].max() + pd.Timedelta(days=1)

# Считаем RFM-метрики по каждому клиенту
rfm = (
    rfm_df.groupby("CUSTOMERNAME")
          .agg(
              recency=("ORDERDATE", lambda x: (snapshot_date - x.max()).days),
              frequency=("ORDERNUMBER", "nunique"),
              monetary=("SALES", "sum")
          )
          .reset_index()
)

# Смотрим первые строки результата
rfm.head()
```

```python
# R score: чем меньше дней с последней покупки, тем лучше
rfm["r_score"] = pd.qcut(
    rfm["recency"],
    q=5,
    labels=[5, 4, 3, 2, 1],
    duplicates="drop"
)

# F score: чем больше заказов, тем лучше
rfm["f_score"] = pd.qcut(
    rfm["frequency"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5],
    duplicates="drop"
)

# M score: чем больше денег потратил клиент, тем лучше
rfm["m_score"] = pd.qcut(
    rfm["monetary"],
    q=5,
    labels=[1, 2, 3, 4, 5],
    duplicates="drop"
)

# Собираем общий RFM-код, например 555, 431, 212
rfm["rfm_score"] = (
    rfm["r_score"].astype(str)
    + rfm["f_score"].astype(str)
    + rfm["m_score"].astype(str)
)

rfm.head()
```

```python
# Простая функция для сегментации клиентов
def rfm_segment(row):
    r = int(row["r_score"])
    f = int(row["f_score"])
    m = int(row["m_score"])

    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"
    elif r >= 4 and f >= 3:
        return "Loyal Customers"
    elif r >= 4 and f <= 2:
        return "New Customers"
    elif r <= 2 and f >= 4 and m >= 4:
        return "At Risk VIP"
    elif r <= 2 and f <= 2:
        return "Lost Customers"
    else:
        return "Regular Customers"

# Применяем функцию к каждому клиенту
rfm["segment"] = rfm.apply(rfm_segment, axis=1)

rfm.head()
```

```python
# Смотрим количество клиентов в каждом сегменте
rfm["segment"].value_counts()
```

```python
# Делаем сводку по сегментам
segment_summary = (
    rfm.groupby("segment")
       .agg(
           customers=("CUSTOMERNAME", "count"),
           avg_recency=("recency", "mean"),
           avg_frequency=("frequency", "mean"),
           total_monetary=("monetary", "sum"),
           avg_monetary=("monetary", "mean")
       )
       .sort_values("total_monetary", ascending=False)
)

segment_summary
```

### RFM Analysis Summary

The RFM analysis segments customers based on three behavioral metrics:

- `Recency`: how recently the customer made a purchase.
- `Frequency`: how many unique orders the customer made.
- `Monetary`: how much revenue the customer generated.

Customers with high R, F, and M scores are the most valuable customers, while customers with low recency and low frequency may require reactivation campaigns.

The phone number formatting issue does not impact this segmentation, because the RFM model uses purchase behavior and customer names, not phone numbers.
